```python
age = int(input("What is your age?"))
if age >= 18:
    country = input("What is your country?")
    if country == "Pakistan":     # nested if
        print("You can vote in Pakistan.")
    else:
        print("You can not vote in Pakistan")
else:
    print("You cannot vote.")
```

```python
age = int(input("What is your age?"))
country = input("What is your country?").capitalize()
if age > 18 and country == "Pakistan":       #combine conditions
    print("Eligible voter")
else:
    print("Not eligible voter")
```
