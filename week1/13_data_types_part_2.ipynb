```python
# Python Collections
# Used to store multiple values in one variable

# 1. List           -> Ordered, Changeable
# 2. Tuple          -> Ordered, Unchangeable
# 3. Set            -> Unordered, Unique values
# 4. Dictionary     -> Key-Value pairs
```

# What is a List?

A list is an ordered, changeable collection.

**✓** Ordered **✓** Mutable (can change) **✓**  Allows duplicates


```python
fruits = ["apple", "banana", "mango", "banana"]
print(fruits)
```

# Accessing Elements

```python
print(fruits[0])  # Accessing the first element
print(fruits[-1])  # Accessing the last element
```

# List Methods (Most Important)

```python
fruits.append("orange")  # Add item
```

```python
fruits.insert(1, "grapes")  # Insert at index
print(fruits)
```

```python
fruits.remove("banana")  # Remove item
print(fruits)
```

```python
fruits.pop()  # Remove last item
print(fruits)
```

```python
fruits.sort()  # Sort list
print(fruits)
```

# Looping Through List

```python
for fruit in fruits:
    print(fruit)
```

# What is a Tuple?
A tuple is an ordered, unchangeable collection.

**✓** Ordered **✗** Cannot modify after creation  


```python
colors = ("red", "green", "blue")
print(colors)
```

# Access Tuple

```python
print(colors[1])
```

# Tuple Properties

```python
# Tuples are immutable (cannot be changed)
colors[0] = "black" # This will cause an error
```

# Tuple Methods

```python
numbers = (1, 2, 3, 2, 4, 2)
print(numbers.count(2))
print(numbers.index(3))
```

# What is a Set?
A set stores unique unordered values.

**✓** No duplicates **✓** Fast operations

```python
my_set = {1, 2, 3, 4, 4, 5}
print(my_set) # No duplicates
```

# Set Methods

```python
my_set.add(6)
my_set.remove(2)
print(my_set)
```

# Set Operations

```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
```

# What is a Dictionary?
Stores data as key : value pairs.

```python
student = {
    "name": "Ali",
    "age": 22,
    "course": "Data Science"
}

print(student)
```

# Access Values

```python
print(student["name"])
print(student.get("age"))
```

# Dictionary Methods

```python
student["age"] = 23     # Update

student["city"] = "Lahore"   #Add

student.pop("course")   # Remove

print(student)
```

# Looping Dictionary

```python
for key, value in student.items():
    print(key, ":", value)
```

# Quick Comparison

```python
# List  -> Ordered, Mutable
# Tuple -> Ordered, Immutable
# Set   -> Unordered, Unique
# Dict  -> Key-Value, Fast Lookup
```
