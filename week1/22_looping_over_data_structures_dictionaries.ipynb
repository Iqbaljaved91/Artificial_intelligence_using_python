```python
student = {"name": "Ali", "age": 20, "grade": "A"}

for key in student:
    print(key)
```

```python
student = {"name": "Ali", "age": 20, "grade": "A"}

for key in student:
    print(student[key])
```

```python
student = {"name": "Ali", "age": 20, "grade": "A"}

for key in student:
    print(key , ":", student[key])
```

```python
student_list = [
    {"name": "Ali", "age": 20, "grade": "A"},
    {"name": "Sara", "age": 22, "grade": "B"},
]
for student in student_list:
    print(f"Name: {student.get('name', '')} | Grade: {student.get('grade', '')}")
```
