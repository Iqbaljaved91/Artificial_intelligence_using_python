```python
print(5 + 3)      # addition
print(10 - 4)     # subtraction
```

```python
print(2 * 6)      # multiply
print(15 / 3)     # divide
print(17 // 3)    # floor divide
```

```python
print(10 % 3)     # remainder
print(2 ** 3)     # exponent
```

```python
print(5 == 5)     # equality
print(7 != 3)     # Inequality
print(10 > 2)     # Greater than
print(4 < 1)      # Less than
```

```python
print(True and False)        # binary "and" operator
print(True or False)         # binary "or" operator
print(not True)              # binary "not" operator
```

```python
x = 10    # simple assignment
print(x)  
x += 5    # same as x = x + 5
print(x)
x -= 3    # same as x = x - 3
print(x)
x *= 2    # same as x = x * 2
print(x)
x /= 4    # same as x = x / 4
print(x)
```
