```python
class Car:
    def drive(self):
        print("The car is moving!")
```

```python
car1 = Car()
car1.drive()
```

```python
class Car:
    color = "red"
    def drive(self):
        print(f"{self.color} car is moving!")
```

```python
car1 = Car()
car1.drive()
```

```python
class Car:
    color = "red"
    def drive(self):
        print(f"{self.color} car is moving!")
    def setColor(self, new_color):
        self.color = new_color
```

```python
car2 = Car()
car2.setColor("green")
car2.drive()
```
