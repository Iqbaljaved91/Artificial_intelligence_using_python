```python
class Animal:
    def speak(self):
        print("Some sound")
```

```python
class Dog(Animal):
    pass
```

```python
dog1 = Dog()
```

```python
dog1.speak()
```

```python
class Dog(Animal):
    def bark(self):
        print("Woof!")
```

```python
childdog = Dog()
```

```python
childdog.bark()
```

```python
childdog.speak()
```
