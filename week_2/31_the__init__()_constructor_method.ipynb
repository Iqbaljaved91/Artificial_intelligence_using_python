```python
class Dog:
    def __init__(self):
        self.name = "Buddy"
    def bark(self):
        print(f"{self.name} says Woof!")
```

```python
dog1 = Dog()
dog1.bark()
```

```python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color
```

```python
car1 = Car("Toyota", "Red")
print(car1.brand)
print(car1.color)
```

```python
car1.color = "Blue"
car2 = Car("Honda", "White")
print(car2.brand)
print(car2.color)
print(car1.color)
```
