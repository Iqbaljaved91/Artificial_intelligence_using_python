```python
num = int(input("Enter a number: "))
print(10 / num)
```

```python
try:
    num = int(input("Enter a number"))
    print(10 / num)
except ZeroDivisionError:
    print("You can't divide by zero!")
except ValueError:
    print("That's not a number!")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
try:
    print("Running code...")
except:
    print("Error found!")
else:
    print("No errors found!")
finally:
    print("Done!")
```
